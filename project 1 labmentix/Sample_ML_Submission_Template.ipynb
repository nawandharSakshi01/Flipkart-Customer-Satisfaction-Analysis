## ***5. Hypothesis Testing***

Hypothesis testing is a statistical method used to determine whether a claim about the data is true or not.

### Hypothetical Statement - 1

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

The customer support channel has a significant impact on customer satisfaction (CSAT) score.

Null Hypothesis (H₀)

There is no significant difference in CSAT scores among different support channels (Inbound, Outcall, and Email).

Alternative Hypothesis (H₁)

There is a significant difference in CSAT scores among different support channels.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import f_oneway

# Separate CSAT scores by support channel
inbound = df1[df1['channel_name'] == 'Inbound']['CSAT Score']
outcall = df1[df1['channel_name'] == 'Outcall']['CSAT Score']
email = df1[df1['channel_name'] == 'Email']['CSAT Score']

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(inbound, outcall, email)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis (H0)")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H0)")
````

##### Which statistical test have you done to obtain P-Value?

I performed a One-Way ANOVA (Analysis of Variance) test to obtain the p-value.

If p-value < 0.05 → Reject the Null Hypothesis (H₀).

If p-value ≥ 0.05 → Fail to Reject the Null Hypothesis (H₀).

##### Why did you choose the specific statistical test?

I chose the One-Way ANOVA test because I wanted to compare the mean CSAT scores across three different support channels (Inbound, Outcall, and Email). ANOVA is the appropriate statistical test when comparing the means of more than two independent groups to determine whether the difference is statistically significant.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Customer satisfaction (CSAT Score) differs across different agent shifts.

Null Hypothesis (H₀):

There is no significant difference in CSAT scores across different agent shifts.

Alternative Hypothesis (H₁):

There is a significant difference in CSAT scores across different agent shifts.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import f_oneway

morning = df1[df1['Agent Shift'] == 'Morning']['CSAT Score']
evening = df1[df1['Agent Shift'] == 'Evening']['CSAT Score']
night = df1[df1['Agent Shift'] == 'Night']['CSAT Score']
split = df1[df1['Agent Shift'] == 'Split']['CSAT Score']

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(morning, evening, night, split)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis (H0)")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H0)")
````

##### Which statistical test have you done to obtain P-Value?

I performed the One-Way ANOVA (Analysis of Variance) test to obtain the p-value.

##### Why did you choose the specific statistical test?

I chose the One-Way ANOVA test because I wanted to compare the mean CSAT scores across four different agent shifts (Morning, Evening, Night, and Split). ANOVA is the appropriate test when comparing the means of more than two independent groups.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Customer satisfaction (CSAT Score) differs across different issue categories

Null Hypothesis (H₀):

There is no significant difference in CSAT scores across different issue categories.

Alternative Hypothesis (H₁):

There is a significant difference in CSAT scores across different issue categories.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import f_oneway

# Create groups based on issue category
groups = [group['CSAT Score'].dropna() for name, group in df1.groupby('category')]

# Perform One-Way ANOVA
f_stat, p_value = f_oneway(*groups)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)

alpha = 0.05

if p_value < alpha:
    print("Decision: Reject the Null Hypothesis (H0)")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H0)")
````

##### Which statistical test have you done to obtain P-Value?

I performed the One-Way ANOVA (Analysis of Variance) test to obtain the p-value.

##### Why did you choose the specific statistical test?

I chose the One-Way ANOVA test because I wanted to compare the mean CSAT scores across multiple issue categories. ANOVA is the appropriate statistical test for comparing the means of more than two independent groups.
